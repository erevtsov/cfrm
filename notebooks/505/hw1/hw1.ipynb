## Problem 1

Consider the random variables $X$ and $Y$ with the following joint distribution:

$$f_{XY}(x,y) = \begin{cases} \frac{1}{Z}(xy^2 + 2xy) & 0 < x, y < 1 \\ 0 & \text{otherwise} \end{cases}$$

where $Z$ is some constant. Calculate the following quantities analytically (**not** by simulation).

### Finding Z

Require $\int_0^1 \int_0^1 \frac{1}{Z}(xy^2 + 2xy)\, dy\, dx = 1$.

Inner integral (over $y$):
$$\int_0^1 (xy^2 + 2xy)\, dy = \left[\frac{xy^3}{3} + xy^2\right]_0^1 = \frac{x}{3} + x = \frac{4x}{3}$$

Outer integral (over $x$):
$$\int_0^1 \frac{4x}{3}\, dx = \frac{4}{3}\cdot\frac{x^2}{2}\Bigg|_0^1 = \frac{2}{3}$$

So $\frac{1}{Z}\cdot\frac{2}{3} = 1 \Rightarrow Z = \frac{2}{3}$, and thus:

$$\boxed{f_{XY}(x,y) = \frac{3}{2}(xy^2 + 2xy), \quad 0 < x,y < 1}$$

### 1.1 Part a: $\mathbb{P}[X + Y < 0.5]$

Integration region: $x > 0,\ y > 0,\ x + y < 0.5$, i.e. $0 \le x \le 0.5$, $0 \le y \le 0.5 - x$.

Inner integral:
$$\int_0^{0.5-x} \frac{3}{2}(xy^2+2xy)\, dy = \frac{3}{2}\left[\frac{xy^3}{3}+xy^2\right]_0^{0.5-x} = \frac{3}{2}\cdot x(0.5-x)^2\left[\frac{0.5-x}{3}+1\right]$$
$$= \frac{1}{2}\,x(0.5-x)^2(3.5-x)$$

Expanding $x(0.5-x)^2(3.5-x)$:
$$= x(0.25 - x + x^2)(3.5-x) = 0.875x - 3.75x^2 + 4.5x^3 - x^4$$

Outer integral:
$$\mathbb{P}[X+Y<0.5] = \int_0^{0.5}\frac{1}{2}(0.875x - 3.75x^2 + 4.5x^3 - x^4)\, dx$$
$$= \frac{1}{2}\left[\frac{7}{16}x^2 - \frac{25}{3}\cdot\frac{x^3}{3} + \frac{9}{4}x^4 - \frac{x^5}{5}\right]_0^{0.5}$$

$$\boxed{\mathbb{P}[X+Y<0.5] \approx 0.00899}$$

### 1.2 Part b: $\mathbb{E}[X]$

Marginal $f_X(x)$:
$$f_X(x) = \int_0^1 \frac{3}{2}(xy^2+2xy)\, dy = \frac{3}{2}\left[\frac{xy^3}{3}+xy^2\right]_0^1 = \frac{3}{2}\left(\frac{x}{3}+x\right) = \frac{3}{2}\cdot\frac{4x}{3} = 2x$$

Expected value:
$$\mathbb{E}[X] = \int_0^1 x\cdot 2x\, dx = \int_0^1 2x^2\, dx = \frac{2}{3}x^3\Bigg|_0^1$$

$$\boxed{\mathbb{E}[X] = \frac{2}{3}}$$

### 1.3 Part c: $\mathrm{Var}[Y]$

$$f_Y(y) = \int_0^1 \frac{3}{2}(xy^2+2xy)\, dx = \frac{3}{2}(y^2+2y)\int_0^1 x\, dx = \frac{3}{2}(y^2+2y)\cdot\frac{1}{2} = \frac{3}{4}(y^2+2y)$$

$$\mathbb{E}[Y] = \int_0^1 y\cdot\frac{3}{4}(y^2+2y)\, dy = \frac{3}{4}\int_0^1(y^3+2y^2)\, dy = \frac{3}{4}\left[\frac{y^4}{4}+\frac{2y^3}{3}\right]_0^1 = \frac{3}{4}\left(\frac{1}{4}+\frac{2}{3}\right) = \frac{3}{4}\cdot\frac{11}{12} = \frac{11}{16}$$

$$\mathbb{E}[Y^2] = \int_0^1 y^2\cdot\frac{3}{4}(y^2+2y)\, dy = \frac{3}{4}\int_0^1(y^4+2y^3)\, dy = \frac{3}{4}\left[\frac{y^5}{5}+\frac{y^4}{2}\right]_0^1 = \frac{3}{4}\left(\frac{1}{5}+\frac{1}{2}\right) = \frac{3}{4}\cdot\frac{7}{10} = \frac{21}{40}$$

$$\mathrm{Var}[Y] = \mathbb{E}[Y^2] - (\mathbb{E}[Y])^2 = \frac{21}{40} - \left(\frac{11}{16}\right)^2 = \frac{21}{40} - \frac{121}{256}$$

$$\boxed{\mathrm{Var}[Y] = \frac{67}{1280}}$$

### 1.4 Part d: $\mathrm{Cov}[X, Y]$

$$\mathrm{Cov}[X,Y] = \mathbb{E}[XY] - \mathbb{E}[X]\mathbb{E}[Y]$$

$$\mathbb{E}[XY] = \int_0^1\int_0^1 xy\cdot\frac{3}{2}(xy^2+2xy)\, dy\, dx = \frac{3}{2}\int_0^1\int_0^1(x^2y^3+2x^2y^2)\, dy\, dx$$

$$\int_0^1(x^2y^3+2x^2y^2)\, dy = x^2\left[\frac{y^4}{4}+\frac{2y^3}{3}\right]_0^1 = x^2\cdot\frac{11}{12}$$

$$\mathbb{E}[XY] = \frac{3}{2}\int_0^1\frac{11}{12}x^2\, dx = \frac{3}{2}\cdot\frac{11}{12}\cdot\frac{1}{3} = \frac{11}{24}$$

$$\mathrm{Cov}[X,Y] = \frac{11}{24} - \frac{2}{3}\cdot\frac{11}{16} = \frac{11}{24} - \frac{22}{48} = \frac{11}{24} - \frac{11}{24}$$

$$\boxed{\mathrm{Cov}[X,Y] = 0}$$

### 1.5 Part e: $\mathrm{Corr}[Y, X]$

$$\mathrm{Corr}[Y,X] = \frac{\mathrm{Cov}[X,Y]}{\sqrt{\mathrm{Var}[X]}\cdot\sqrt{\mathrm{Var}[Y]}} = \frac{0}{\cdots}$$

$$\boxed{\mathrm{Corr}[Y,X] = 0}$$

### 1.6 Part f: $\mathbb{P}[X > 0.5 \mid Y > 0.5]$

$$\mathbb{P}[X>0.5 \mid Y>0.5] = \frac{\mathbb{P}[X>0.5 \cap Y>0.5]}{\mathbb{P}[Y>0.5]}$$

$$\int_{0.5}^1\frac{3}{2}(xy^2+2xy)\, dy = \frac{3}{2}\left[\frac{xy^3}{3}+xy^2\right]_{0.5}^1 = \frac{3}{2}\left[\frac{4x}{3} - \left(\frac{x}{24}+\frac{x}{4}\right)\right] = \frac{3}{2}\cdot\frac{25x}{24} = \frac{25x}{16}$$

$$\mathbb{P}[X>0.5\cap Y>0.5] = \int_{0.5}^1\frac{25x}{16}\, dx = \frac{25}{16}\cdot\frac{x^2}{2}\Bigg|_{0.5}^1 = \frac{25}{32}\cdot\frac{3}{4} = \frac{75}{128}$$

$$\mathbb{P}[Y>0.5] = \int_{0.5}^1\frac{3}{4}(y^2+2y)\, dy = \frac{3}{4}\left[\frac{y^3}{3}+y^2\right]_{0.5}^1 = \frac{3}{4}\cdot\frac{25}{24} = \frac{25}{32}$$

$$\mathbb{P}[X>0.5\mid Y>0.5] = \frac{75/128}{25/32} = \frac{75}{128}\cdot\frac{32}{25}$$

$$\boxed{\mathbb{P}[X>0.5\mid Y>0.5] = \frac{3}{4}}$$

### 1.7 Part g: $\mathbb{E}[X \mid Y = 0.25]$

Conditional density $f_{X|Y}(x|y)$:
$$f_{X|Y}(x|y) = \frac{f_{XY}(x,y)}{f_Y(y)} = \frac{\frac{3}{2}x(y^2+2y)}{\frac{3}{4}(y^2+2y)} = 2x$$

The conditional density is independent of $y$, consistent with $\mathrm{Cov}[X,Y]=0$ and the factorization $f_{XY}=f_X\cdot f_Y$.

$$\mathbb{E}[X\mid Y=0.25] = \int_0^1 x\cdot 2x\, dx = \frac{2}{3}$$

$$\boxed{\mathbb{E}[X\mid Y=0.25] = \frac{2}{3} = \mathbb{E}[X]}$$

## Problem 2

Use Monte Carlo integration to estimate
$$\int_{-2}^{3} 3x^2 + 2x\, dx$$
(The exact answer is 40.)

Approach: Sample $X \sim U(-2, 3)$ (interval length $b-a=5$), then
$$\int_{-2}^{3} f(x)\, dx \approx (b-a)\cdot\frac{1}{n}\sum_{i=1}^n f(X_i)$$

```python
n = 100_000
a, b = -2, 3

X = np.random.uniform(a, b, size=n)
f = 3*X**2 + 2*X

estimate = (b - a) * np.mean(f)
print(f"Monte Carlo estimate: {estimate:.4f}")
print(f"Exact answer:         40")
```

## Problem 3
Let $U \sim U(0,1)$ and $X = 1 - U$. Confirm analytically and by Monte Carlo:
$$\mathbb{E}[X] = \frac{1}{2}, \quad \mathrm{Var}[X] = \frac{1}{12}, \quad \mathrm{Cov}[U,X] = -\frac{1}{12}$$

### Analytical derivations

Since $X = 1-U$ has $f_X(x) = 1$ on $(0,1)$:

$$\mathbb{E}[X] = \int_0^1 x\, dx = \frac{1}{2}x^2\Bigg|_0^1 = \frac{1}{2}$$

$$\mathbb{E}[X^2] = \int_0^1 x^2\, dx = \frac{1}{3}x^3\Bigg|_0^1 = \frac{1}{3}$$

$$\mathrm{Var}[X] = \mathbb{E}[X^2] - (\mathbb{E}[X])^2 = \frac{1}{3} - \frac{1}{4} = \frac{1}{12}$$

$$\mathrm{Cov}[U,X] = \mathbb{E}[UX] - \mathbb{E}[U]\mathbb{E}[X] = \mathbb{E}[U(1-U)] - \frac{1}{2}\cdot\frac{1}{2}$$
$$= \mathbb{E}[U] - \mathbb{E}[U^2] - \frac{1}{4} = \frac{1}{2} - \frac{1}{3} - \frac{1}{4} = \frac{6-4-3}{12} = -\frac{1}{12}$$

```python
n = 100_000
U = np.random.uniform(0, 1, size=n)
X = 1 - U

print(f"E[X]      = {np.mean(X):.5f}   (exact: 0.5)")
print(f"Var[X]    = {np.var(X):.5f}   (exact: {1/12:.5f})")
print(f"Cov[U,X]  = {np.cov(U, X)[0,1]:.5f}  (exact: {-1/12:.5f})")
```

## Problem 4

### 4.1 Part a

Substitution:
$$u = \frac{x}{1+x} \implies x = \frac{u}{1-u}, \quad dx = \frac{1}{(1-u)^2}\, du$$

When $x=0 \Rightarrow u=0$; when $x\to\infty \Rightarrow u\to 1$.

Derivation:
$$u(1+x) = x \implies u + ux = x \implies u = x - ux = x(1-u) \implies x = \frac{u}{1-u}$$

$$\int_0^\infty xe^{-x}\, dx = \int_0^1 \frac{u}{1-u}\cdot e^{-u/(1-u)}\cdot\frac{1}{(1-u)^2}\, du = \int_0^1 \frac{u}{(1-u)^3}\,e^{-u/(1-u)}\, du$$

$$= \mathbb{E}\left[\frac{U}{(1-U)^3}\,e^{-U/(1-U)}\right], \quad U\sim U(0,1)$$

```python
n = 100_000
U = np.random.uniform(0, 1, size=n)

g = (U / (1 - U)**3) * np.exp(-U / (1 - U))
print(f"Part a estimate: {np.mean(g):.5f}  (exact: 1)")
```

### 4.2 Part b

Substitution:
$$u = 1 - e^{-x} \implies e^{-x} = 1-u \implies x = -\ln(1-u), \quad \frac{du}{dx} = e^{-x} \implies dx = \frac{1}{e^{-x}}\, du = \frac{1}{1-u}\, du$$

When $x=0 \Rightarrow u=0$; when $x\to\infty \Rightarrow u\to 1$.

$$\int_0^\infty xe^{-x}\, dx = \int_0^1 (-\ln(1-u))\cdot(1-u)\cdot\frac{1}{1-u}\, du = \int_0^1 -\ln(1-u)\, du$$

$$= \mathbb{E}[-\ln(1-U)], \quad U\sim U(0,1)$$

```python
n = 100_000
U = np.random.uniform(0, 1, size=n)

g = -np.log(1 - U)
print(f"Part b estimate: {np.mean(g):.5f}  (exact: 1)")
```

### 4.3 Part c

Since $X \sim \mathrm{Exp}(1)$ has PDF $f_X(x) = e^{-x}$ for $x \ge 0$:

$$\int_0^\infty xe^{-x}\, dx = \int_0^\infty x\cdot f_X(x)\, dx = \mathbb{E}[X]$$

where $X \sim \mathrm{Exp}(1)$.

```python
n = 100_000
X = np.random.exponential(1, size=n)

print(f"Part c estimate: {np.mean(X):.5f}  (exact: 1)")
```

## Problem 5

$X = \max\{U_1, U_2\}$

Let $U_1$ and $U_2$ be i.i.d. $U(0,1)$. Define $X = \max\{U_1, U_2\}$.

### 5.1 Part a 

$\theta = \mathbb{P}[X \le 0.5]$ by hand

$$\theta = \mathbb{P}[\max\{U_1,U_2\} \le 0.5] = \mathbb{P}[U_1 \le 0.5 \cap U_2 \le 0.5]$$

Since $U_1$ and $U_2$ are independent:
$$= \mathbb{P}[U_1 \le 0.5]\cdot\mathbb{P}[U_2 \le 0.5] = 0.5 \times 0.5$$

$$\boxed{\theta = 0.25}$$

### 5.2 Part b

```python
n = 100_000
u1 = np.random.uniform(0, 1, size=n)
u2 = np.random.uniform(0, 1, size=n)

X = np.maximum(u1, u2)
theta_hat = np.mean(X <= 0.5)

print(f"Monte Carlo estimate of theta: {theta_hat:.5f}")
print(f"Exact theta:                   0.25")
```

### 5.3 Part c: Why `u2 = 1 - u1` doesn't work

The approach: generate `u1 = np.random.uniform(0,1,size=n)` and set `u2 = 1 - u1`.

Why it doesn't work: Although $1-U_1 \sim U(0,1)$ marginally, $U_2 = 1-U_1$ is not independent of $U_1$. They are perfectly negatively correlated. The problem requires $U_1 \perp U_2$.

With $U_2 = 1-U_1$:
$$X = \max\{U_1,\, 1-U_1\} \ge 0.5 \quad \text{always}$$
- If $U_1 < 0.5$: $\max = 1-U_1 > 0.5$
- If $U_1 \ge 0.5$: $\max = U_1 \ge 0.5$

So $X > 0.5$ with probability 1, giving $\hat\theta \approx 0$ instead of $0.25$.

```python
n = 100_000
u1 = np.random.uniform(0, 1, size=n)
u2 = 1 - u1  # WRONG: not independent

X_bad = np.maximum(u1, u2)
theta_bad = np.mean(X_bad <= 0.5)

print(f"Wrong estimate (u2=1-u1): {theta_bad:.5f}  (should be ~0.25, gets ~0)")
```
