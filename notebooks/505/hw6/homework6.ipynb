# Problem 1

Consider the integral

$$\theta = \int_{0}^{1}\cos(x)\,\textrm{d}x$$

## Part a

Estimate this integral using direct Monte Carlo simulation.  First use a pilot study with a sample size of at least $100$ to estimate how many samples you would need to ensure that $P(|\hat{\theta}_{N} - \theta| \leq 0.01) \geq 0.95$, then use a full-sized simulation with the appropriate number of samples.  Report the required number of samples, your estimate of the integral, a confidence interval for your estimate and the exact error.

## Part b

Repeat part (a), but use antithetic variates for both the pilot study and the full simulation.

## Part c

How much did the necessary sample size drop from part (a) to part (b)?  (Be sure you're comparing equivalent values: For example, $N$ vs $N$ instead of $2N$ vs $N$.)

### Solution
#### Part a


```python
N0 = 1_000
N_max = 10_000_000
zdelta = sps.norm.ppf(1 - 0.05 / 2)
epsilon = 0.01

u = np.random.uniform(0, 1, size=N0)
x_pilot = np.cos(u)
theta_N0 = np.mean(x_pilot)
var_N0 = np.var(x_pilot)
print("Pilot study: ")
print(f"Pilot mean = {theta_N0:.5f}")
print(f"Pilot var = {var_N0:.5f}")

N_est = int(np.ceil(zdelta ** 2 * var_N0 / epsilon ** 2))
N = min(N_est, N_max)
print(f"Estimate of required sample size = {N_est}")

u = np.random.uniform(0, 1, size=N)
x = np.cos(u)

theta = np.mean(x)
var = np.var(x)
lb = theta - zdelta * np.sqrt(var / N)
ub = theta + zdelta * np.sqrt(var / N)
print("\nFull size study: ")
print(f"Estimate = {theta:.5f}")
print(f"Confidence interval: [{lb:.5f}, {ub:.5f}]")
print(f"Exact error: {abs(theta - np.sin(1)):.5f}")
```

#### Part b

```python
N0 = 1_000
N_max = 10_000_000
zdelta = sps.norm.ppf(1 - 0.05 / 2)
epsilon = 0.01

N0_av = int(N0 / 2)
u = np.random.uniform(0, 1, size=N0_av)
x_pilot = (np.cos(u) + np.cos(1 - u)) / 2

theta_N0 = np.mean(x_pilot)
var_N0 = np.var(x_pilot)
print("Pilot study: ")
print(f"Pilot mean = {theta_N0:.5f}")
print(f"Pilot var = {var_N0:.5f}")

N_est = int(np.ceil(zdelta ** 2 * var_N0 / epsilon ** 2))
N_av = min(N_est, N_max)
print(f"Estimate of required antithetic pairs = {N_est}")

u = np.random.uniform(0, 1, size=N_av)
x = (np.cos(u) + np.cos(1 - u)) / 2

theta = np.mean(x)
var = np.var(x)
lb = theta - zdelta * np.sqrt(var / N_av)
ub = theta + zdelta * np.sqrt(var / N_av)
print("\nFull size study: ")
print(f"Estimate = {theta:.5f}")
print(f"Confidence interval: [{lb:.5f}, {ub:.5f}]")
print(f"Exact error: {abs(theta - np.sin(1)):.5f}")
```

#### Part c

The confidence intervals in parts (a) and (b) are comparable in width -- both are roughly $\pm 0.01$ -- because both pilot studies were calibrated to the same target: $P(|\hat{\theta}_N - \theta| \leq 0.01) \geq 0.95$. The advantage of antithetic variates is not a narrower CI but a much smaller required sample size to reach the same precision.

The variance of the direct estimator is $\text{Var}[\cos(U)] \approx 0.02$, while the variance of the antithetic estimator is $\text{Var}\!\left[\frac{\cos(U)+\cos(1-U)}{2}\right] \approx 0.001$, a roughly 20x reduction. Since $N \propto \sigma^2 / \epsilon^2$, the required number of samples drops by the same factor: from roughly $N \approx 770$ in part (a) to roughly $N_\text{pairs} \approx 40$ pairs (80 function evaluations) in part (b), approximately a 10x reduction in total computation.

#### Part d

Antithetic variates reduce variance because the paired estimator $Y_i = \frac{f(U_i)+f(1-U_i)}{2}$ satisfies

$$\text{Var}[Y_i] = \frac{\text{Var}[f(U)]}{2}\bigl(1 + \text{Corr}[f(U),\,f(1-U)]\bigr)$$

Large variance reduction requires $\text{Corr}[f(U), f(1-U)] \ll 0$.

Problem 1 -- $\cos(x)$: $\cos$ is monotone decreasing on $[0,1]$ (traversing only ~16% of one period). For any monotone $f$, $f(U)$ and $f(1-U)$ are negatively correlated: when $U$ is large, $1-U$ is small, so $f(U)$ is small while $f(1-U)$ is large. The strong negative correlation (~$-0.9$) drives the large variance reduction.

Problem 2 -- $\cos(6x)$: Since $6/(2\pi) \approx 0.955$, $\cos(6u)$ completes almost exactly one full period on $[0,1]$.

$$\text{Corr}[\cos(6U),\,\cos(6(1-U))] = \frac{\sin(6)/12 + \cos(6)/2 - (\sin(6)/6)^2}{\text{Var}[\cos(6U)]} \approx 0.956$$

The correlation is strongly positive because $\cos(6u)$ and $\cos(6(1-u)) = \cos(6-6u)$ traverse nearly the same period and tend to have the same sign at the same time. With $\text{Corr} \approx 0.956$, the variance ratio per pair is $(1+0.956)/2 \approx 0.978$ -- barely any reduction.

# Problem 3

In this problem, you will explore several methods to estimate

$$\theta = \mathbb{E}\left[X^2\right]$$

where $X\sim N(0, 4)$ is a normal random variable with $\mu = 0$ and $\sigma^2 = 4$.

## Part a

Use the direct approach with $N = 100,000$ samples.  Report your estimate of $\theta$ along with a confidence interval and the standard error.

## Part b

Use antithetic variates with $N = 50,000$ samples.  (Each sample should be the average of two antithetic variates - one generated through $Z$ and the other through $-Z$.  Report your estimate of $\theta$ along with a confidence interval and the standard error.  By what percentage did you reduce the standard error from part (a)?  Explain why this happens.

## Part c

Use $X$ as a control variate and use $N = 100,000$ samples.  Report your estimate of $\theta$ along with a confidence interval and the standard error.  By what percentage did you reduce the standard error from part (a)?  Explain why this happens.

### Solution
#### Part a

```python
N_a = 100_000
zdelta = sps.norm.ppf(1 - 0.05 / 2)

X = np.random.normal(0, 2, size=N_a)  # N(0, sigma^2=4) -> sigma=2
f_a = X ** 2

theta_a = np.mean(f_a)
var_a = np.var(f_a)
se_a = np.sqrt(var_a / N_a)
lb_a = theta_a - zdelta * se_a
ub_a = theta_a + zdelta * se_a

print(f"Estimate       = {theta_a:.5f}")
print(f"Standard error = {se_a:.5f}")
print(f"95% CI         = [{lb_a:.5f}, {ub_a:.5f}]")
```

#### Part b

```python
N_b = 50_000

Z = np.random.normal(0, 1, size=N_b)
X_pos = 2 * Z        # ~ N(0, 4)
X_neg = -2 * Z       # antithetic: -X ~ N(0, 4)
f_b = (X_pos ** 2 + X_neg ** 2) / 2

theta_b = np.mean(f_b)
var_b = np.var(f_b)
se_b = np.sqrt(var_b / N_b)
lb_b = theta_b - zdelta * se_b
ub_b = theta_b + zdelta * se_b
pct = (se_b - se_a) / se_a * 100

print(f"Estimate       = {theta_b:.5f}")
print(f"Standard error = {se_b:.5f}")
print(f"95% CI         = [{lb_b:.5f}, {ub_b:.5f}]")
print(f"SE change vs part (a): {pct:+.1f}%")
```

The standard error increases by roughly $\sqrt{2} \approx 41\%$ rather than decreasing.

The antithetic variate of $X$ is $-X$, but $f(-X) = (-X)^2 = X^2 = f(X)$. Since $f$ is an even function and the distribution of $X$ is symmetric about 0, the two variates in each pair are identical: $\text{Corr}[f(X), f(-X)] = 1$. The paired estimator $Y_i = \frac{f(X_i) + f(-X_i)}{2} = f(X_i)$ carries no new information, so there is zero variance reduction from the pairing. With $N = 50{,}000$ pairs instead of $100{,}000$ direct samples, the standard error simply scales as $1/\sqrt{N}$, increasing by a factor of $\sqrt{100{,}000 / 50{,}000} = \sqrt{2}$.

#### Part c

```python
N_c = 100_000

X = np.random.normal(0, 2, size=N_c)
f_c_raw = X ** 2
Z = X  # control variate; E[X] = 0

c_hat = -np.cov(f_c_raw, Z)[0, 1] / np.var(Z)
f_c = f_c_raw + c_hat * Z  # E[Z] = 0, so no mean-shift needed

theta_c = np.mean(f_c)
var_c = np.var(f_c)
se_c = np.sqrt(var_c / N_c)
lb_c = theta_c - zdelta * se_c
ub_c = theta_c + zdelta * se_c
pct_c = (se_c - se_a) / se_a * 100

print(f"c_hat          = {c_hat:.5f}")
print(f"Estimate       = {theta_c:.5f}")
print(f"Standard error = {se_c:.5f}")
print(f"95% CI         = [{lb_c:.5f}, {ub_c:.5f}]")
print(f"SE change vs part (a): {pct_c:+.1f}%")
```

The standard error is essentially unchanged from part (a), and $\hat{c} \approx 0$.

The optimal coefficient is $c^* = -\text{Cov}[X^2, X] / \text{Var}[X]$. For $X \sim N(0, 4)$:

$$\text{Cov}[X^2, X] = \mathbb{E}[X^3] - \mathbb{E}[X^2]\,\mathbb{E}[X] = \mathbb{E}[X^3] = 0$$

since $X^3$ is an odd function of a distribution symmetric about 0, so all odd moments vanish. Therefore $c^* = 0$ and $\text{Corr}[X^2, X] = 0$: $X$ is completely uncorrelated with $X^2$ and provides no variance reduction. The standard error satisfies $\text{se}_c = \text{se}_a \sqrt{1 - \rho^2} = \text{se}_a$, i.e., no improvement over part (a).

# Problem 4

Consider an asset whose value $S_t$ is governed by the SDE

$$\textrm{d}S_t = \mu S_t\,\textrm{d}t + \sigma(t) S_t\,\textrm{d}W_t$$

where $\mu$ is a constant and

$$\sigma(t) = 0.2 + 0.1\sin(t)$$

Use the parameters $\mu = 0.1$, $T = 5$ and $S_0 = 1$.

For each part, implement your method using at least 100 time steps and at least $10,000$ samples.

## Part a

Estimate $\mathbb{E}[S_T]$ using direct simulation.  That is, use the Euler method to simulate $S_t$ up until time $T$ at least $10,000$ times and then average the final values.  Report your estimate along with a 95% confidence interval and the standard error.

## Part b

Let $X_t$ be the value of a hypothetical asset governed by

$$\textrm{d}X_t = \mu X_t\,\textrm{d}t + \overline{\sigma}X_t\,\textrm{d}W_t$$

where

$$\overline{\sigma} = \frac{1}{T}\int_{0}^{T}\sigma(t)\,\textrm{d}t$$

Notice that $X_t\sim\textrm{GBM}(\mu, \overline{\sigma})$ and so we know the distribution and expected value of $X_T$.

Estimate $\mathbb{E}[S_T]$ using $X_T$ as a control variate.  (This means that you should use the Euler method to simulate both $S_t$ and $X_t$ with *the same* noise and use the resulting $X_T$ as a control).  You can estimate the optimal value of $c$ from your full sample (i.e., you don't have to calculate it by hand or do a separate pilot study).  Report your estimate along with a 95% confidence interval and the standard error. How much did this reduce the standard error from part (a)?

### Solution
#### Part a

```python
T = 5
s0 = 1
mu = 0.1
def sigma(t):
    return 0.2 + 0.1*np.sin(t)

N = 100_000
N_steps = 101
t = np.linspace(0, T, N_steps)
dt = t[1] - t[0]
s = np.zeros((N, N_steps))
s[:, 0] = s0

for k in range(N_steps-1):
    z = np.random.normal(0, 1, size=N)
    s[:, k+1] = s[:, k] + mu * s[:, k] * dt + sigma(t[k]) * s[:, k] * np.sqrt(dt) * z

s_t = s[:, -1]
theta = np.mean(s_t)
var = np.var(s_t)
se = np.sqrt(var / N)
lb = theta - zdelta * se
ub = theta + zdelta * se

print(f"Estimate       = {theta:.5f}")
print(f"Standard error = {se:.5f}")
print(f"95% CI         = [{lb:.5f}, {ub:.5f}]")
```

#### Part b

```python
sigma_bar = (0.2 * T - 0.1 * np.cos(T) + 0.1) / T
EX_T = s0 * np.exp(mu * T)

s_curr = np.full(N, float(s0))
x_curr = np.full(N, float(s0))

for k in range(N_steps - 1):
    z = np.random.normal(0, 1, size=N)
    s_curr = s_curr + mu * s_curr * dt + sigma(t[k]) * s_curr * np.sqrt(dt) * z
    x_curr = x_curr + mu * x_curr * dt + sigma_bar   * x_curr * np.sqrt(dt) * z

s_T = s_curr
x_T = x_curr

c_hat = -np.cov(s_T, x_T)[0, 1] / np.var(x_T)
f_c = s_T + c_hat * (x_T - EX_T)

theta_c = np.mean(f_c)
se_c = np.sqrt(np.var(f_c) / N)
lb_c = theta_c - zdelta * se_c
ub_c = theta_c + zdelta * se_c
pct = (se_c - se) / se * 100

print(f"sigma_bar      = {sigma_bar:.5f}")
print(f"c_hat          = {c_hat:.5f}")
print(f"Estimate       = {theta_c:.5f}")
print(f"Standard error = {se_c:.5f}")
print(f"95% CI         = [{lb_c:.5f}, {ub_c:.5f}]")
print(f"SE change vs part (a): {pct:+.1f}%")
```

$\bar{\sigma}$ is computed analytically from the closed-form antiderivative of $\sigma(t) = 0.2 + 0.1\sin(t)$:

$$\bar{\sigma} = \frac{1}{T}\int_0^T \sigma(t)\,dt = \frac{1}{T}\bigl[0.2t - 0.1\cos(t)\bigr]_0^T = \frac{1.1 - 0.1\cos(5)}{5} \approx 0.2143$$

Both $S_t$ and $X_t$ are simulated with the same Brownian increments $z$ at each step. Because $X_t$ uses constant $\bar{\sigma}$ while $S_t$ uses time-varying $\sigma(t)$, the two paths differ only in their local volatility -- they are driven by identical noise, making $X_T$ and $S_T$ highly correlated. This strong correlation is what makes $X_T$ an effective control variate.

Since $X_T \sim \text{GBM}(\mu, \bar{\sigma})$ with $X_0 = S_0 = 1$, its expected value is known exactly: $\mathbb{E}[X_T] = e^{\mu T}$. The corrected estimator $\hat{\theta}_c = \frac{1}{N}\sum_i \bigl(S_T^{(i)} + \hat{c}(X_T^{(i)} - e^{\mu T})\bigr)$ is unbiased for $\mathbb{E}[S_T]$, and the large SE reduction reflects how tightly $S_T$ and $X_T$ co-move.

# Problem 5

In the previous homework, you estimated the value of both an arithmetic and a geometric Asian option with the parameters

$$\begin{aligned}
S_0 = 50,\, K = 50,\, T = 1,\, n = 5,\, r = 0.05,\, \sigma = 0.2.
\end{aligned}$$

You also confirmed the formula for the exact value of the geometric Asian option.  In this problem, you will estimate the value of the same arithmetic Asian option with the method of control variates.  You are welcome to cite or re-use code/results from homework 5.

## Part a

Estimate the value of the arithmetic Asian option $A_A$ by using the value of the geometric Asian option as a control.  Use a sample size of at least $100,000$.  In other words, for each sample you should simulate one path of $\{S_t\}$ and use that path to calculate the time-discounted payoff of both the arithmetic and geometric options (using the same parameters as above).  Because you know the expected value of the payoff of the geometric version (from homework 5), you can use it as a control.  You can estimate the optimal value of $c$ from your full sample (i.e., you don't have to calculate it by hand or do a separate pilot study).

## Part b

How much did this reduce the standard error from the direct version in the previous homework?

### Solution
#### Part a

```python
# Parameters (same as hw5 Problems 2 & 3)
N = 100_000
K = 50
T = 1.0
s_0 = 50
r = 0.05
sigma = 0.2
n = 5
d_t = T / n

# Closed-form C_G (reproduced from hw5 Problem 2d)
var_term = (2*n**2 + 3*n + 1) * d_t / (6*n)
d_cf = (np.log(s_0 / K) + (r - sigma**2/2) * (d_t*(n+1)/2)) / (sigma * np.sqrt(var_term))
exp_term = np.exp(-r*(d_t*(n-1)/2) - (sigma**2/2) * (d_t*(n**2-1)/(6*n)))
C_G = s_0 * exp_term * sps.norm.cdf(d_cf + sigma*np.sqrt(var_term)) - K*np.exp(-r*T)*sps.norm.cdf(d_cf)

# Simulate N paths at monitoring times t_1,...,t_n (vectorized version of hw5 approach)
paths = np.zeros((N, n))
s_prev = np.full(N, float(s_0))
for i in range(n):
    Z = np.random.normal(0, 1, size=N)
    s_curr = s_prev * np.exp((r - sigma**2/2) * d_t + sigma * np.sqrt(d_t) * Z)
    paths[:, i] = s_curr
    s_prev = s_curr

# Discounted payoffs for arithmetic and geometric options on the same paths
S_A = np.mean(paths, axis=1)
S_G = np.exp(np.mean(np.log(paths), axis=1))  # geometric mean via log-average
payoff_A = np.exp(-r*T) * np.maximum(S_A - K, 0)
payoff_G = np.exp(-r*T) * np.maximum(S_G - K, 0)

# Control variate correction using C_G as the known expectation of payoff_G
c_hat = -np.cov(payoff_A, payoff_G)[0, 1] / np.var(payoff_G)
payoff_c = payoff_A + c_hat * (payoff_G - C_G)

theta_c = np.mean(payoff_c)
se_c = np.sqrt(np.var(payoff_c) / N)
lb_c = theta_c - zdelta * se_c
ub_c = theta_c + zdelta * se_c

print(f"C_G (closed-form) = {C_G:.6f}")
print(f"c_hat             = {c_hat:.5f}")
print(f"A_A estimate      = {theta_c:.6f}")
print(f"Standard error    = {se_c:.6f}")
print(f"95% CI            = [{lb_c:.6f}, {ub_c:.6f}]")
```

#### Part b

```python
se_direct = np.sqrt(np.var(payoff_A) / N)
pct = (se_c - se_direct) / se_direct * 100

print(f"Direct SE (hw5 approach): {se_direct:.6f}")
print(f"Controlled SE:            {se_c:.6f}")
print(f"SE reduction:             {pct:+.1f}%")
```
